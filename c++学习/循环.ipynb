```python
a = 0
while a < 10:
    a +=1
    if a%2 == 0:
        continue
    print(a)  #注意缩进几次，print要放在while里面
```

```python
row = 1
while row < 10:
    col = 1
    while col <= row:
        print(f'{col} x {row} = {row*col}',sep=' ')
        col += 1
    row += 1
```
